```python
def prioritas_tiket(jenis, pengguna, urgensi, berhenti):
    """
    Menentukan prioritas tiket berdasarkan rule IF-THEN
    """

    # R1
    if jenis == "Server" and berhenti:
        return "KRITIS"

    # R2
    elif pengguna >= 100 and urgensi == "Tinggi":
        return "TINGGI"

    # R3
    elif jenis == "Printer" and urgensi == "Rendah":
        return "RENDAH"

    # R4
    elif jenis == "PC" and 10 <= pengguna < 100:
        return "SEDANG"

    # R5
    else:
        return "SEDANG"
```

```python
jenis = input("Jenis gangguan (Server/PC/Printer): ")
pengguna = int(input("Jumlah pengguna terdampak: "))
urgensi = input("Tingkat urgensi (Tinggi/Sedang/Rendah): ")

status = input("Apakah layanan berhenti? (ya/tidak): ")

berhenti = status.lower() == "ya"

hasil = prioritas_tiket(jenis, pengguna, urgensi, berhenti)

print("\n===== HASIL ANALISIS =====")
print("Jenis Gangguan :", jenis)
print("Pengguna       :", pengguna)
print("Urgensi        :", urgensi)
print("Layanan Stop   :", berhenti)
print("---------------------------")
print("PRIORITAS      :", hasil)
```

```python
data_uji = [
    ("Server",230,"Tinggi",True),
    ("Server",20,"Sedang",True),
    ("Printer",5,"Rendah",False),
    ("PC",45,"Sedang",False),
    ("PC",150,"Tinggi",False),
    ("Server",120,"Tinggi",False),
    ("Printer",60,"Tinggi",False),
    ("PC",8,"Rendah",False),
    ("Server",300,"Tinggi",True),
    ("PC",70,"Sedang",False)
]

print("No | Jenis | Prioritas")
print("-"*35)

for i,d in enumerate(data_uji,1):
    hasil = prioritas_tiket(*d)
    print(f"{i:2} | {d[0]:8} | {hasil}")
```
